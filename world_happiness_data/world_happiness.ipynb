## 9. Question-wise analysis

### Question 1. How has average life evaluation changed over time?

```python
# Calculate the average life evaluation for each year
yearly_average = (
    data.groupby("year")["life_evaluation_3-year_average"]
    .mean()
    .reset_index()
)

# Display the result
print(yearly_average)
```

```python
# Plot average life evaluation over the years
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=yearly_average,
    x="year",
    y="life_evaluation_3-year_average",
    marker="o"
)

plt.title("Average Life Evaluation Over Time")
plt.xlabel("Year")
plt.ylabel("Average Life Evaluation")

plt.show()
```

## Interpretation
- The average life evaluation shows an overall increasing trend from 2011 to 2025.
- It decreased slightly between 2012 and 2016, reaching its lowest point around 2016.
- From 2017 to 2021, the average life evaluation increased steadily.
- There was a small decline from 2021 to 2023.
- From 2023 to 2025, the average increased sharply, reaching the highest level in 2025 (around 5.65).
- Overall, the chart suggests that global average life evaluation has improved over the period, despite some year-to-year fluctuations.

### Question 2. Which countries experienced the largest improvement or decline?

```python
# Find the first year in the dataset
first_year = data["year"].min()

# Find the latest year in the dataset
latest_year = data["year"].max()

print("First year:", first_year)
print("Latest year:", latest_year)
```

```python
# Get country scores from the first year
first = data[data["year"] == first_year][
    ["country_name", "life_evaluation_3-year_average"]
]
first

# Rename the score column
first = first.rename(
    columns={
        "life_evaluation_3-year_average": "first_score"
    }
)

print(first.head())
```

```python
# Get country scores from the latest year
latest = data[data["year"] == latest_year][
    ["country_name", "life_evaluation_3-year_average"]
]
latest

# Rename the score column
latest = latest.rename(
    columns={
        "life_evaluation_3-year_average": "latest_score"
    }
)

print(latest.head())
```

```python
# Merge first-year and latest-year scores
change = pd.merge(
    first,
    latest,
    on="country_name",
    how="inner"
)

print(change.head())
```

```python
# Calculate the change in life evaluation
change["change"] = (
    change["latest_score"] - change["first_score"]
)

# Sort from largest improvement to largest decline
change = change.sort_values(
    "change",
    ascending=False
)

print(change.head(10))
```

```python
# Top 10 countries with the largest improvement
improved = change.head(10)

print(improved)
```

```python
# Plot top 10 countries with the largest improvement
plt.figure(figsize=(10, 6))

sns.barplot(
    data=improved,
    x="change",
    y="country_name"
)

plt.title("Top 10 Countries with Largest Improvement in Life Evaluation")
plt.xlabel("Change in Life Evaluation")
plt.ylabel("Country")

plt.show()
```

```python
# Top 10 countries with the largest decline
declined = change.tail(10).sort_values("change")

print(declined)
```

```python
# Plot top 10 countries with the largest decline
plt.figure(figsize=(10, 6))

sns.barplot(
    data=declined,
    x="change",
    y="country_name"
)

plt.title("Top 10 Countries with Largest Decline in Life Evaluation")
plt.xlabel("Change in Life Evaluation")
plt.ylabel("Country")

plt.show()
```

## Key Insights Summary
- Life evaluation generally improved from 2011 to 2025.
- Social support has the strongest relationship with life evaluation (0.70).
- GDP per capita (0.68) and healthy life expectancy (0.66) also show strong positive relationships.
- Countries experienced different levels of improvement and decline over time.
- Generosity shows almost no relationship with life evaluation (0.04).
