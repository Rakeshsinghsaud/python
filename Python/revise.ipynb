### comprehension
- short and concise ways

```python
#list comprehension:[expression for item in iterable if condition]
even_num=[]
for i in range(100):
    if i %2==0:
        even_num.append(i)

print(even_num)
```

```python
result=[i**2 for i in range(100) if i%2==0]
print(result)
```

```python
result=[ (i,j) for i in range(1,11) for j in range(1,5)]
print(result)
```

```python
#set comprehension
num={ i for i in range(100) if i%2!=0}
print(num)
```

```python
#Dictionary
result={i:i**2 for i in range(10)}
print(result)
```
