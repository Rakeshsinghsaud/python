```python
#*args takes any number of positional arguments, args is tuple so any number\n
#  - argument we pass  get stored in tuple.
def sum(*args):
    print(args)
    z = 0
    for i in args:
        z+=i
    return z
    
```

```python
# **kwargs takes any number of keyword arguments.
def greet(**kwargs):
    
    print(kwargs)
    for k,v in kwargs.items():
        print(f"{k}:{v}")
```

### Getter and setter Method:
- ​Getter and setter methods are used to read and update private data safely.
- Getter method reads private data.​
- ​Setter method updates private data.​
- ​​Setter can validate data before updating.​
- ​​​This protects the object from invalid values.​

- syntax:
-      def get_value(self):
-           return self.__value
-      def set_value(self,value):
-           self.__value=value

```python
class student:
    def __init__(self):
        self.__marks=0
    def get_marks(self):#getter
        return self.__marks
    def set_marks(self,marks):#setter
        if marks>=0 and marks<=100:
           self.__marks=marks
        else:
            print("invalid")
s=student()
s.set_marks(90)
print(s.get_marks())
```

```python
#@property is used to access a method like an attribute.

class student:
    def __init__(self):
        self.__marks=0

    @property
    def marks(self):
         return self.__marks
    
    @marks.setter
    def marks(self,marks):
        if marks>0 and marks<=100:
         self.__marks=marks
        else:
            print("invalid")

s=student()

s.marks=90
print(s.marks)
```

### Inheritance
- Inheritance allows one class to reuse the properties and methods of another class.​
- ​Inheritance supports code reuse.​
- ​​ ​The parent class is also called base class or superclass.​
- ​  ​The child class is also called derived class or subclass.​
- ​​ ​The child class can use parent class attributes and methods.​
- ​​ ​The child class can also define its own attributes and methods.​
- ​​ ​The child class can override parent methods.​​ ​Inheritance represents an​​ is-a ​​relationship.​

```python
#Exampe of  single inheritance:one child class inherits from one parent class.​
class Animal: #parent class
    def eat(slef):
        print("eating")
class Dog(Animal):#child class
    def bark(self):
        print("barking")
d=Dog()
d.eat()
d.bark()
```

```python
#Multilevel inheritance: child class inherites the property and method from  parents class forming chain.
class Animal:
    def eat(self):
        print("eating")

class Dog(Animal):
     def bark(self):
         print("barking")

class puppy(Dog):
    def sweep(self):
        print("sweeping")

p=puppy()
p.sweep()
p.bark()
p.eat()
```

```python
#Hierarchical Inheritance:multiple child classes inherit from one parent class.​

class Animal:
    def eat(self):
        print("eating")


class Dog(Animal):
    def bark(self):
        print("barking")


class cat(Animal):
    def meow(self):
        print("meowing")

d=Dog()
c=cat()
d.eat()
d.bark()
c.eat()
c.meow()
```

```python
#Multiple Inheritance:one child class inherits from more than one parent class.​

class father:
    def father_skill(self):
        print("Gardening")

class mother:
    def mother_skill(self):
        print("painting")

class child(father,mother):
    def child_skill(self):
        print("coding")

c=child()
c.father_skill()
c.mother_skill()
c.child_skill()
```

### Hybrid Inheritance
- combination of two or more than two type of inheritance.
- It can include multiple, multilevel, or hierarchical inheritance together.​
- ​It is powerful but can become complex.​

```python
class person:
    def show_person(self):
        print("person")


class student(person):
    def show_student(self):
        print("student")


class employee(person):
    def show_employee(self):
        print("employee")


class working_student(student,employee):
        def show_working(self):
             print("working student")

w=working_student()
w.show_person()
w.show_student()
w.show_employee()
w.show_working()
```

### Method overloading
- method name of both child and parent class is same.
-

```python
class Animal:
      def sound(self):
         print("Animal sound")

class Dog(Animal):
      def sound(self):
         super().sound() # calls the parent version first 
         print("Barking")

d=Dog()
d.sound()
```

### Polymorphism
- Methods or operation behaves differently in different situations.
- Polymorphism means many forms.
- The same method name can  works differently in different classes.
- ​Python supports polymorphism through:​
- ​1.​ ​Method overriding​
- ​2.​ ​Duck typing​
- 3.​ ​Operator overloading​

```python
#method overriding

class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Barking")

d=Dog()
d.sound()
```

```python
#duck typing

class Dog:
    def sound(self):
        print("bark")

class Cat:
    def sound(self):
        print("meow")

animals=[Dog(),Cat()]

for animal in animals:
    animal.sound()
```

```python
# operator overloading

class box:
    def __init__(self,length):
        self.length=length

    def __add__(self,other):
        return (self.length + other.length)

box1=box(20)
box2=box(30)
print(box1 + box2)
```
