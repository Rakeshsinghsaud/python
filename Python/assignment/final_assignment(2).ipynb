### Q1. Movie Theatre Seat Booking
A small theatre has **10 seats**. Customers book seats one by one until the show is full or they stop booking.

**Guided Steps:**
- Create a function `book_seat(booked_seats, seat_number)` that adds a seat number to a list, only if it is not already booked and is between 1 and 10
- Use a `while` loop to keep asking the user to enter a seat number to book
- If the user enters `0`, stop the loop using `break`
- If the seat number is invalid (not between 1-10) or already booked, print a message and use `continue` to ask again
- After every successful booking, print how many seats are still available
- If all 10 seats get booked, print `"Show is house full"` and stop the loop automatically

```python
def book_seat(booked_seats,seat_number):
    if seat_number>=1 and seat_number<=10 and seat_number is not booked_seats:
        booked_seats.append(seat_number)
        return True
    
    return False

booked_seats=[]
while len(booked_seats)<10:
    seat_number=int(input("enter seat number between 1 to 10"))
    if seat_number==0:
        break
    if seat_number <1 or seat_number >10:
        print("invalid")
        continue
    if seat_number is  booked_seats:
        print("seat is alredy booked")
        continue
    book_seat(booked_seats,seat_number)
    available_seat=10 -len(booked_seats)
    print("seats are booked successfully")
    print(f"seat availables are {available_seat}")
if len(booked_seats)==10:
    print("house full")


```

### Detailed Guidance for the Final Project

Use this section if you get stuck on any step — it's a nudge, not a full solution.

**Step 1 (the `MenuItem` class):**

**Step 3 (updating the cart inside a function):**
- This is the same idea as the shopping cart question earlier — if you pass a *list* into a function and `.append()` to it, the original list outside the function changes too, because lists are mutable and passed by object reference.

**Step 4 (the ordering loop):**

**Step 5 (error handling):**

**Step 6 (the receipt with zip and enumerate):**
- If you have two related lists — ordered item names and their prices — `zip()` lets you walk through both together instead of tracking two separate index variables. Wrap the result in `enumerate()` if you also want a running item number for the receipt.

**Step 9 (the discount lambda):**
