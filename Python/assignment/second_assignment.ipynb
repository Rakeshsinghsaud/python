### A1. ATM Withdrawal Checker
**Scenario:** You are building a mini ATM system. A user enters the amount they want to withdraw. Your program must decide if the withdrawal is allowed.

**Guided Steps:**
1. Assume the account `balance = 5000`.
2. Take the withdrawal amount as input and convert it to an integer.
3. If the amount is **not a multiple of 100**, print `"Amount must be in multiples of 100"`.
4. Else if the amount is **greater than the balance**, print `"Insufficient balance"`.
5. Else, deduct the amount from the balance and print the **new balance**.


```python
balance=5000
amount=int(input("enter the amount that you want to withdraw"))
if amount%100!=0:
    print("Amount must be in the multiple of 100.")
elif amount > balance:
    print("Insufficient balance.")
else:
    balance=balance - amount
    print(f"new balance is {balance}")
```

### B3. Password Retry System
**Scenario:** A login system allows a user up to 3 attempts to enter the correct password.

**Guided Steps:**
1. Set `correct_password = "python123"` and `attempts = 3`.
2. Use a `while` loop that runs as long as `attempts > 0`.
3. Inside the loop, take the password as input.
4. If it matches, print "Access granted" and use `break` to exit the loop.
5. If it doesn't match, decrease `attempts` by 1 and print how many attempts remain.
6. If the loop ends without success, print "Account locked".


```python
correct_password= "python123"
attempt=3
while attempt>0:
    password=input("enter a password.")
    if password==correct_password:
        print("Access granted")
        break
    else:
        attempt-=1
        print(f"remaining attempt is {attempt}")
    if attempt==0:
       print("Account locked")
```

## Part C: Break and Continue

### C1. Attendance Register (skip absentees)
**Scenario:** A teacher has a list of students marked present (`1`) or absent (`0`). She wants to print only the names of students who are present, and stop as soon as she reaches a student named "STOP" (a special marker for end of list).

**Guided Steps:**
1. Create two lists: `students = ["Aman", "Riya", "Sam", "STOP", "Neha"]` and `status = [1, 0, 1, 0, 1]`.
2. Loop through the students list using a `for` loop with `range(len(students))` so you can access both lists using the same index.
3. If the current student is `"STOP"`, use `break` to stop the loop immediately.
4. If the student's status is `0` (absent), use `continue` to skip printing them.
5. Otherwise, print the student's name as present.


```python
students=["Aman","Riya","Sam","STOP","Neha"]
status=[1,0,1,0,1]
for i in range(len(students)):
    if students[i]=="STOP":
        break
    if status[i]==0:
        continue
    print(students[i], "is present.")
```

### C2. Prime Number Finder (for OTP generation)
**Scenario:** A security system wants to find the first prime number greater than a number you give it, to use as part of a one-time code.

**Guided Steps:**
1. Take a starting number `n` as input.
2. Use a `while` loop that keeps increasing `n` by 1 each time.
3. For each `n`, use a `for` loop from 2 to `n-1` to check for factors.
4. If a factor divides `n` evenly, use `break` to stop checking (it's not prime) and move to the next number.
5. Use a flag variable to track whether `n` is prime; once you find one, print it and `break` the outer loop too.


```python
n=int(input("enter a number"))
while n<=20:
    n+=1
    prime=True
    for i in range(2,n):
        if n%i==0:
            prime=False
            break

    if prime==True:
        print(f"The prime number is {n}")
```

### C3. Online Order Validator
**Scenario:** An e-commerce site is processing a batch of orders. Orders with a negative quantity are invalid and should be skipped (they were entered by mistake). Processing should stop completely if an order quantity of `-999` (a system error code) appears.

**Guided Steps:**
1. Create a list: `orders = [5, 2, -1, 8, -999, 3, 10]`.
2. Loop through the list with a `for` loop.
3. If the order quantity is `-999`, print "System error encountered, stopping processing" and `break`.
4. If the order quantity is negative (but not -999), print "Invalid order skipped" and `continue`.
5. Otherwise, print "Processing order of quantity: X" and add it to a running `total_items` count.
6. After the loop, print the total valid items processed.


```python
orders=[5,2,-1,8.-999,3,10]
total_items=0
for order in orders:
    if order==-999:
        print("system error encountered.")
        break
    if order<0:
        print("invalid order skipped")
        continue
    print(f"processing order of quantity {order}")
    total_items+=order
print(f"total valid items processed {total_items}")

```
